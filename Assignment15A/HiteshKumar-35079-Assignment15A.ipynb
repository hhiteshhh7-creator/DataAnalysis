## Q1: What is the average fare by service type?

```python
average_fare = df.groupby('service_type')['fare'].mean()
print(average_fare.round(2))
```

## Q2: Which city has the most trips?

```python
city_trips = df.groupby('city')['trip_id'].count()
city_trips = city_trips.sort_values(ascending=False)
print(city_trips)
print('City with the most trips:', city_trips.index[0])
```

## Q3: What is the average distance per trip?

```python
average_distance = df['distance_km'].mean()
print('Average distance per trip:', round(average_distance, 2), 'km')
```

## Q4: Compare trip ratings by gender.

```python
average_rating = df.groupby('gender')['trip_rating'].mean()
print(average_rating.round(2))
print('Male riders have the higher average trip rating in this dataset.')
```

## Q5: Make a bar chart of service type vs. average fare.

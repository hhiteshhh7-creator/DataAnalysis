## Q1: What is the most preferred tea type?

```python
tea_counts = df.groupby('tea_type')['student_id'].count()
tea_counts = tea_counts.sort_values(ascending=False)
print(tea_counts)
```

**Answer:** Green tea is the most frequently recorded tea type (29 students).

## Q2: What is the average number of cups per gender?

```python
average_cups = df.groupby('gender')['cups_per_day'].mean()
print(average_cups)
```

**Answer:** Males average 3.00 cups per day, while females average 2.60 cups per day.

## Q3: Make a bar chart of snack choices.

```python
snack_counts = df.groupby('snack_choice')['student_id'].count()
snack_counts = snack_counts.sort_values(ascending=False)
print(snack_counts)

plt.figure(figsize=(8, 5))
plt.bar(snack_counts.index, snack_counts.values)
plt.xlabel('Snack Choice')
plt.ylabel('Number of Students')
plt.title('Snack Choices of Students')
plt.show()
```

**Answer:** Samosa is the most popular snack with 40 students, followed by Chips (31) and Biscuit (29).

## Q4: Compare tea preference by age group.

```python
tea_by_age = df.groupby(['age_group', 'tea_type'])['student_id'].count()
print(tea_by_age)
```

**Answer:** Green tea leads among ages 18–20 (13) and 24–26 (11), while Black tea leads among ages 21–23 (9).

## Q5: Count the number of students who don't drink tea.
